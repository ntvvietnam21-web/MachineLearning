```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

np.random.seed(42)
x = 2 * np.random.rand(100, 1)
nhieu_goc = np.random.randn(100, 1)

cac_muc_nhieu = [0.5, 1.0, 2.0, 5.0]

plt.figure(figsize=(16, 4))

for i, muc_nhieu in enumerate(cac_muc_nhieu, 1):
    y = 4 + 3 * x + muc_nhieu * nhieu_goc
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

    model = LinearRegression()
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)

    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    print(f"Mức nhiễu = {muc_nhieu} | MSE = {mse:.4f} | R^2 = {r2:.4f}")

    plt.subplot(1, 4, i)
    plt.scatter(x_test, y_test, color="blue", alpha=0.6, label="Thực tế")
    plt.plot(x_test, y_pred, color="red", linewidth=2, label="Dự đoán")
    plt.title(f"Nhiễu = {muc_nhieu}\nMSE={mse:.2f}, R2={r2:.2f}")
    plt.xlabel("x")
    if i == 1:
        plt.ylabel("y")
    plt.legend()

plt.tight_layout()
plt.show()
```
